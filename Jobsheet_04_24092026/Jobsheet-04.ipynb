```python
import pandas as pd
import numpy as np

df = pd.read_csv('sales_electroprima2.csv')
df.head()
df.shape
```

```python
p_malang = (df['branch'] == 'Malang').mean()
print('p(branch == Malang) =', round(p_malang, 4))
```

```python
condition = (df['product_category'] == 'Laptop') & \
            (df['branch'] == 'Jakarta')
p_joint = condition.mean()
print('P(Laptop and Jakarta) =', round(p_joint, 4))
```

```python
malang = df[df['branch'] == 'Malang']
p_conditional = (malang['satisfaction_rating'] == 5).mean()

print('P(rating=5 | Malang) =', round(p_conditional, 4))
print('Number of Malang transactions:', len(malang))
```

```python
np.random.seed(42)
sim_result = np.random.choice(df['branch'], size=1000)

for n in [10, 100, 500, 1000]:
    proportion = (sim_result[:n] == 'Malang').mean()
    print(f'n={n:5d}  Malang proportion = {proportion:.4f}')

print('P(Malang) from the original data =', round(p_malang, 4))
```

```python
table = pd.crosstab(df['branch'] == 'Malang',
                    df['satisfaction_rating'] == 5, margins=True)
table.index = ['Not Malang', 'Malang', 'Total']
table.columns = ['Rating != 5', 'Rating = 5', 'Total']
print(table)
```

```python
p_surabaya = (df['branch'] == 'Surabaya').mean()
print("P(branch == 'Surabaya') =", round(p_surabaya, 4))

p_smart_bandung = ((df['product_category'] == 'Smartphone') & (df['branch'] == 'Bandung')).mean()
print("P(Smartphone and Bandung) =", round(p_smart_bandung, 4))
```

```python
jakarta = df[df['branch'] == 'Jakarta']
p_rating5_jakarta = (jakarta['satisfaction_rating'] == 5).mean()

print('P(rating=5 | Jakarta) =', round(p_rating5_jakarta, 4))
print('P(rating=5 | Malang) =', round(p_conditional, 4))
```

```python
p_acc_surabaya = ((df['product_category'] == 'Accessories') & (df['branch'] == 'Surabaya')).mean()
print("P(Accessories and Surabaya) =", round(p_acc_surabaya, 4))

np.random.seed(42)
sim_product = np.random.choice(df['product_category'], size=1000)

p_acc_true = (df['product_category'] == 'Accessories').mean()

for n in [10, 100, 500, 1000]:
    prop = (sim_product[:n] == 'Accessories').mean()
    print(f'n={n:5d}  Accessories proportion = {prop:.4f}')

print("P(Accessories) dari data asli =", round(p_acc_true, 4))
```
